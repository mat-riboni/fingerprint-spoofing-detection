# Gaussian Mixture Model

```python
from gaussian_mixture_utils import *
import numpy as np
from utils import *

D, L = load("train_data.txt")
(DTR, LTR), (DVAL, LVAL) = split_db_2to1(D, L)
pi_target = 0.1

D0 = DTR[:, LTR == 0]
D1 = DTR[:, LTR == 1]
```

```python
best_GMM = []
best_min_DCF = np.inf

for i in range(1, 5):
    GMM0 = LBG(D0, 2**i)
    GMM1 = LBG(D1, 2**i)
    _, logdens0 = logpdf_GMM(DVAL, GMM0)
    _, logdens1 = logpdf_GMM(DVAL, GMM1)
    log_scores = logdens1 - logdens0

    predictions = (log_scores > 0).astype(int)
    accuracy = np.sum(predictions == LVAL) / len(LVAL)

    minDCF = compute_min_dcf(log_scores, LVAL, pi_target, 1, 1)

    if minDCF < best_min_DCF:
        best_min_DCF = minDCF
        best_GMM = [GMM0, GMM1]

    print(f"Model with {2**i} components | Accuracy: {accuracy} | Min DCF: {minDCF}")
```
